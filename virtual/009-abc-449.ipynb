## B - Deconstruct Chocolate (6:55)

`-` $1$번 쿼리가 주어졌다고 해보자. 기본적으로 $RW$만큼 존재하나 여태까지 입력된 $2$번 쿼리에 의해 줄어든 너비를 차감하여 고려하면 된다

`-` $2$번 쿼리도 마찬가지로 처리하면 되며 전체 알고리즘의 시간 복잡도는 $O(Q)$이다. $H,W \le 100$이라 나이브하게 $O(HWQ)$에 풀 수도 있다

```python
def solution():
    H, W, Q = map(int, input().split())
    k1 = k2 = 0
    for _ in range(Q):
        op, t = input().split()
        t = int(t)
        if op == "1":
            print(t * (W - k2))
            k1 += t
        else:
            print(t * (H - k1))
            k2 += t


solution()

# input
# 7 9 5
# 2 4
# 1 3
# 2 1
# 2 1
# 1 3
```

## C - Comfortable Distance (35:47)

`-` $j$를 순회하며 가능한 $i$의 개수를 세자. $i$는 $j$로부터 최소 $L$만큼 떨어지고 최대 $R$만큼 떨어진 연속된 구간으로 표현된다. 투 포인터로 구간이 $1$칸씩 오른쪽으로 이동할 때마다 변하는 알파벳의 빈도수를 갱신하면 된다

```python
from collections import defaultdict


def solution():
    N, L, R = map(int, input().split())
    S = input().rstrip()
    counts = defaultdict(int)
    answer = 0
    for j in range(L, N):
        l, r = j - R, j - L
        counts[S[r]] += 1
        if l > 0:
            counts[S[l - 1]] -= 1
        answer += counts[S[j]]
    print(answer)


solution()

# input
# 6 2 4
# aabcba
```

## D - Make Target 2 (82:13)

`-` 나이브는 $O(4 \times 10^{12})$로 TLE이다

`-` 색상은 두 좌표 중 절댓값이 더 큰 값에만 영향을 받음에 주목하자. $x$를 고정시키고 $D \le y \le U$에 대해 검은 점의 개수를 셀 것이다. 이때 축을 기준으로 점의 분포가 대칭이므로 절댓값을 씌워서 생각하면 편하다

`-` $f(x, y)$를 $(x,1), (x,2), \dots, (x,y)$ 중 검은 점의 개수로 정의하자. $1 \le p \le x$에 대해 $\max(x, p) = x$이므로 이 구간 내 검은 점의 개수는 $\min(x,y) \cdot I(\text{$x$ is even})$이다. $x < p \le y$인 구간 내 검은 점의 개수는 $\left\lfloor\frac{y-x}{2}\right\rfloor + I(\text{$(y - x)$ is odd, $x + 1$ is even})$이다. 둘을 적절히 취합해 반환하면 된다

`-` $D$와 $U$에 관계에 따라 적절히 $f$를 활용하면 고정된 $x$에 대해 검은 점의 개수를 계산할 수 있다

`-` $D=U$이면 $I(\text{$\max(x,y)$ is even})$이 검은 점 개수이다. 둘 중 하나가 $0$이라면 $f(x, y) + I(\text{$x$ is even})$이 검은 점 개수이다 ($f$는 $y=0$일 때는 제외하고 계산하기에 따로 셈해줘야 함)

`-` $D > 0$이거나 $U < 0$이라면 구간 합 계산하는 것처럼 셈하자. 이때 $x$축은 구간에 속하지 않으니 무시하자. 이제 남은 경우는 $U > 0, D < 0$으로 각각에 대해 검은 점의 개수를 센 뒤 합하면 된다. 그리고 $x$축을 고려하자

`-` 고정된 $x$마다 검은 점의 개수를 $O(1)$에 셀 수 있으므로 전체 알고리즘의 시간 복잡도는 $O(2 \times 10^6)$이다

```python
def f(x, y):
    x, y = abs(x), abs(y)
    x_is_even = x % 2 == 0
    if y <= x:
        return y * x_is_even
    return x * x_is_even + (y - x) // 2 + ((y - x) % 2 == 1 and not x_is_even)


def solution():
    L, R, D, U = map(int, input().split())
    answer = 0
    for x in range(L, R + 1):
        x_is_even = x % 2 == 0
        if D == U:
            answer += x_is_even
        elif D == 0:
            answer += f(x, U) + x_is_even
        elif U == 0:
            answer += f(x, D) + x_is_even
        elif D > 0:
            answer += f(x, U) - f(x, D - 1)
        elif U < 0:
            answer += f(x, D) - f(x, U + 1)
        else:
            answer += f(x, D) + f(x, U) + x_is_even
    print(answer)


solution()

# input
# -4 3 1 3
```

## E - A += v (Upsolving)

`-` 충분한 연산(최대 $N(M-1)$)을 거치면 $1 \sim M$의 빈도수는 $A$의 최빈값의 빈도수와 같아진다. 그 이후부턴 정렬된 $A$를 통째로 추가하는 것과 같다. 물론 시뮬레이션 마지막 부근엔 온전한 $A$가 아닌 부분 배열을 추가하겠지만 $X \le 10^{18} < 10^{100}$이므로 신경쓰지 않아도 된다

`-` $X \le N$인 경우 간단하게 인덱싱만 하면 된다. $X$가 충분한 연산보다 크다면 사이클을 고려해 계산하면 된다. $X$가 $N$보다 크고 충분한 연산 횟수보다 작다면 어떻게 해야 될까?

`-` 연산은 $1 \sim M$ 중 $A$에 최소로 등장하는 수를 추가하고 만약 여러 개가 존재하면 작은 것부터 넣는다. 그러고 나면 추가한 수는 빈도수가 $1$ 증가했으니 무시되고 $2$번째로 작은 수가 $A$에 추가된다. 이를 반복하면 결국엔 최소 빈도수를 가지는 수들을 정렬한 뒤 $A$에 추가하는 것과 같다

`-` 연산을 계속하다보면 원소의 최소 빈도수는 증가하며 최소 빈도수 집합도 누적됨에 집중하자. 이를 이용하기 위해 쿼리를 정렬해서 $X$가 작은 것부터 처리할 것이다. 빈도수가 같은 것끼리 그룹으로 묶고 빈도수를 기준으로 그룹을 정렬하자. 그룹 내 원소도 정렬해야 된다

`-` $1$번째 그룹이 $2$번째 그룹과 병합되려면 $(f_{2} - f_{1}) s_1$번의 연산이 필요하다 ($f$는 빈도수, $s$는 크기). 여태까지 적용한 연산 횟수 $c$를 기억하자. 쿼리가 주어지면 $X$번의 연산을 거쳤을 때의 $A$를 재현해야 된다. 그러기 위해 다음 그룹과 병합이 가능할 때가지 병합하자. 이는 $c + (f_{2} - f_{1}) s_1 < X - N$을 만족할 때까지 병합을 수행해서 실현할 수 있다. 이때 정렬이 유지되야 하며 원소 추가가 빨라야 하니 세그먼트 트리를 사용하자

`-` 이제 다음 그룹과 병합되려면 많은 연산이 필요하다 (주어진 $X$로는 불가능). 그럼 $1$번째 그룹을 충분히 복제하여 나열한 배열에서 $X - N - c$번째 원소가 쿼리의 답이 된다. 그리고 이는 사이클을 고려해서 계산하면 $1$번째 그룹의 $((X - N - c - 1) \bmod s_1) + 1$번째 원소이다

`-` 이를 모든 쿼리에 대해 수행하면 되며 전체 알고리즘의 시간 복잡도는 최악의 경우 $O\left(N + \sqrt{N}\log \sqrt{N} + Q\log Q + M\log M\right) = O(N + Q\log Q + M\log M)$이다

```python
from collections import Counter, defaultdict


class OrderStatisticTree:
    _root = 1

    def __init__(self, max_value_or_counts):
        is_int = isinstance(max_value_or_counts, int)
        n = max_value_or_counts + 1 if is_int else len(max_value_or_counts)
        self._size = 1 << n.bit_length()
        self._tree = [0] * (self._size << 1)
        if not is_int:
            self._build(max_value_or_counts)

    def _build(self, array):
        for i, a in enumerate(array, start=self._size):
            self._tree[i] = a
        for i in range(self._size - 1, 0, -1):
            self._tree[i] = self._tree[i << 1] + self._tree[i << 1 | 1]

    @property
    def size(self):
        return self._tree[self._root]

    def count(self, x):
        return self._tree[x + self._size]

    def add(self, x, delta):
        i = x + self._size
        while i:
            self._tree[i] += delta
            i >>= 1

    def find_kth(self, k):
        i = 1
        while i < self._size:
            i <<= 1
            if self._tree[i] < k:
                k -= self._tree[i]
                i |= 1
        return i - self._size

    def bisect_right(self, x):
        i = x + self._size
        f_x = 0
        while i:
            if ~i & 1:
                f_x += self._tree[i]
                i -= 1
            i >>= 1
        return f_x


def solution():
    N, M = map(int, input().split())
    A = list(map(int, input().split()))
    Q = int(input())
    queries = [(int(input()), i) for i in range(Q)]
    queries.sort()
    count2array = defaultdict(list)
    for a, count in Counter(A + [*range(1, M + 1)]).items():
        count2array[count - 1].append(a)
    groups = [(count, sorted(array)) for count, array in count2array.items()]
    groups.sort(reverse=True)
    tree = OrderStatisticTree(M)
    f_prev, group = groups.pop()
    for g in group:
        tree.add(g, 1)
    c = 0
    answers = [0] * Q
    for x, i in queries:
        if x <= N:
            answers[i] = A[x - 1]
            continue
        x -= N
        while groups and c + (groups[-1][0] - f_prev) * tree.size < x:
            f, group = groups.pop()
            c += (f - f_prev) * tree.size
            for g in group:
                tree.add(g, 1)
            f_prev = f
        k = ((x - c - 1) % tree.size + 1)
        answers[i] = tree.find_kth(k)
    print("\n".join(map(str, answers)))


solution()

# input
# 3 3
# 1 1 2
# 8
# 1
# 2
# 3
# 4
# 5
# 6
# 7
# 8
```
