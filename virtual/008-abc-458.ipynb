## B - Count Adjacent Cells (4:27)

`-` 각 칸마다 인접한 칸이 몇 개인지 세면 된다. 나는 그냥 순회해서 풀었다 (이게 실수할 가능성이 더 적으며 시간 복잡도도 동일하다)

```python
from itertools import product


def solution():
    H, W = map(int, input().split())
    matrix = [[0] * W for _ in range(H)]
    drc = [(0, -1), (0, 1), (-1, 0), (1, 0)]
    for r, c in product(range(H), range(W)):
        k = 0
        for dr, dc in drc:
            nr, nc = r + dr, c + dc
            is_within_grid = 0 <= nr < H and 0 <= nc < W
            if not is_within_grid:
                continue
            k += 1
        matrix[r][c] = k
    for row in matrix:
        print(*row)


solution()

# input
# 3 3
```

## C - C Stands for Center (9:18)

`-` C의 인덱스를 $i$라 하자 (단, $i$는 $1$부터 시작)

`-` 이를 중심으로 하는 문자열은 $\min(i, N - i + 1)$개 존재한다. 각 C마다 독립이므로 경우의 수를 전부 더하면 정답이다

```python
def solution():
    S = input().rstrip()
    N = len(S)
    answer = 0
    for i, s in enumerate(S, start=1):
        if s != "C":
            continue
        answer += min(i, N - i + 1)
    print(answer)


solution()

# input
# ABCBC
```

## D - Chalkboard Median (12:57)

`-` 웰노운 Running Median 문제이다. 매 쿼리마다 배열을 수정하는 건 $O\left(Q^2 \right)$으로 비효율적이다. 대신 두 개의 힙을 사용하자

`-` 현재 중앙값이 $m$일 때 $m$ 이하의 수들을 관리하는 최대 힙과 $m$ 이상의 수들을 관리하는 최소 힙을 고려하자. 최대 힙은 원소 $i$개를 관리하고 최소 힙은 원소 $i-1$개를 관리하여 최대 힙의 top이 $m$이 되도록 할 것이다

`-` 새로운 두 수 $A,B$가 들어오면 우선 $A$를 최대 힙에 넣자. 그리고 두 힙의 크기를 같게 하기 위해 최대 힙의 top을 최소 힙에 추가할 것이다. 이 과정에서 힙 간의 대소 관계도 정상화된다. 마지막으로 $B$를 최소 힙에 넣은 뒤 최소 힙의 top을 최대 힙에 추가하면 된다. 그럼 최대 힙의 top이 중앙값을 가리키게 된다

`-` 힙에 원소를 추가하고 top을 삭제하는 건 $O(\log Q)$이므로 전체 알고리즘의 시간 복잡도는 $O(Q\log Q)$이다

```python
import heapq


def solution():
    X = int(input())
    Q = int(input())
    max_heap = [-X]
    min_heap = []
    answers = [0] * Q
    for i in range(Q):
        A, B = map(int, input().split())
        heapq.heappush(max_heap, -A)
        heapq.heappush(min_heap, -heapq.heappop(max_heap))
        heapq.heappush(min_heap, B)
        heapq.heappush(max_heap, -heapq.heappop(min_heap))
        answers[i] = -max_heap[0]
    print("\n".join(map(str, answers)))


solution()

# input
# 5
# 3
# 2 3
# 1 2
# 8 9
```

## E - Count 123 (Upsolving)

`-` 제약이 없으면 같은 것이 있는 순열로 딸깍에 풀 수 있다. 이때의 정답은 $\frac{(X_1 + X_2 + X_3)!}{X_1! X_2! X_3!}$이다. 물론 이건 꿈일 뿐...

`-` 임의의 수열 $A$에서 $i$번째 원소까지만 고려한 부분 수열에 대해 $i$번째 원소와 남은 $1,2,3$의 개수가 동일하면 $i+1$번째 원소를 정하는데 있어 동일한 상태로 취급할 수 있다

`-` 그럼 DP로 풀 수 있지만 상태 공간이 $O(X_1 X_2 X_3)$로 너무 커서 안 된다. 제약 때문에 되지 않을까 싶어도 안 된다. $1$과 $2$만 있다고 생각하면 제약이 없으므로 $O(X_1 X_2)$이고 이는 TLE이다

`-` 이를 해결하기 위해선 뭔가 수학적인 관찰이 필요해 보인다. $2$는 배치하는 데 제약이 없는 것에 집중하자

`-` $1$과 $3$을 임의로 배치해보자. 그럼 $2$는 어디에 배치해야 될까? 조건을 만족하기 위해선 인접한 $1$과 $3$ 사이에 적어도 하나 이상의 $2$를 배치해야만 한다. 그리고 남은 $2$는 자유롭게 배치하면 된다

`-` 이제 문제는 $1$과 $3$이 인접한 횟수가 $1 \sim X_2$가 되도록 $1$과 $3$을 적절하게 배치하고 남은 $2$를 배치하는 경우의 수를 구하는 것과 같아졌다

`-` $1$을 $k_1$개의 그룹으로 분할하자. $3$을 $k_3$개의 그룹으로 분할하자. 이때 이들을 번갈아 배치하고 그 사이에 $2$를 배치할 것이므로 $|k_1 - k_3| \le 1$을 만족해야 한다. 또한, $1$과 $3$이 인접한 횟수 $m(= k_1 + k_3 - 1)$에 대해 $m \le X_2$도 만족해야 한다 (당연히 $0 < k_1\le X_1,\, 0 < k_3 \le X_3$)

`-` $1$을 $k_1$개의 그룹으로 분할하는 경우의 수를 구해보자. 각 그룹의 크기를 $a_1, a_2,\dots, a_{k_1}$이라 하면 $a_1 + a_2 + \cdots + a_{k_1} = X_1$을 만족해야 한다. 이때 각 그룹의 크기는 $0$보다 커야하므로 $a'_1 + a'_2 + \cdots + a'_{k_1} = X_1 - k_1$을 만족하는 경우의 수를 구하는 것과 같다. 이는 중복 조합으로 계산하면 $_{k_1}H_{X_1 - k_1} = _{X_1 - 1}C_{\,k_1 - 1}$이다

`-` 마찬가지의 방식으로 계산하면 $3$을 $k_3$개의 그룹으로 분할하는 경우의 수는 $_{k_3}H_{X_3 - k_3} = _{X_3 - 1}C_{\,k_3 - 1}$이다

`-` 이제 $2$를 배치해보자. $2$는 제약이 없어 이미 배치된 수들의 사이와 양 끝에 배치할 수 있으므로 총 $X_1 + X_3 + 1$개의 자리가 있다. 이때 $1$과 $3$이 인접하는 $m$개의 자리에는 $2$를 적어도 하나 배치해야만 하니 이를 고려해 경우의 수를 중복 조합으로 나타내면 $_{X_1 + X_3 + 1}H_{X_2 - m} = _{X_1 + X_2 + X_3 - m}C_{\,X_1 + X_3}$이다

`-` 따라서 $k_1,k_3$가 주어졌을 때 $1,2,3$을 배치하는 경우의 수는 $_{X_1 - 1}C_{\,k_1 - 1} \times _{X_3 - 1}C_{\,k_3 - 1} \times _{X_1 + X_2 + X_3 - m}C_{\,X_1 + X_3}$이다

`-` 이때 $k_1 = k_3$인 경우 $2$를 무시하고 생각해보면 수열 $A$가 $1$로 시작해도 되고 $3$으로 시작해도 되므로 결괏값에 $2$를 곱해주자

`-` 조합은 팩토리얼을 미리 $O(X_1 + X_2 + X_3)$에 전처리 해두면 이후엔 모듈러 역원만 사용해 $O(\log \text{MOD})$에 계산할 수 있으며 가능한 $k_1, k_3$ 쌍의 수는 $O(\min(X_1, X_3))$이다

`-` 따라서 전체 알고리즘의 시간 복잡도는 $O(X_1 + X_2 + X_3 + \min(X_1, X_3) \cdot \log \text{MOD})$이다

```python
from functools import partial


def _nCr(n, r, factorials, mod):
    return (factorials[n] * pow(factorials[r] * factorials[n - r], mod - 2, mod)) % mod


def solution():
    mod = 998244353
    X1, X2, X3 = map(int, input().split())
    n = X1 + X2 + X3
    factorials = [1] * (n + 1)
    for x in range(1, n + 1):
        factorials[x] = (x * factorials[x - 1]) % mod
    nCr = partial(_nCr, factorials=factorials, mod=mod)
    answer = 0
    for k1 in range(1, X1 + 1):
        for k3 in [k1 - 1, k1, k1 + 1]:
            m = k1 + k3 - 1
            if not (0 < k3 <= X3) or X2 < m:
                continue
            count = nCr(X1 - 1, k1 - 1) * nCr(X3 - 1, k3 - 1) * nCr(n - m, X1 + X3) 
            if k1 == k3:
                count *= 2
            answer += count
            answer %= mod
    print(answer)


solution()

# input
# 2 2 1
```

`-` 참고로 `partial`을 쓰면 느려지지만 가독성 때문에 사용했다 (AC 맞는 데 문제 없음)
