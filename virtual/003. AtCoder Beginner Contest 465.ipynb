## A - Supermajority (0:57)

`-` 실수 오차 방지를 위해 $3A \ge 2B$인지 판단하면 된다

```python
def solution():
    A, B = map(int, input().split())
    if 3 * A > 2 * B:
        print("Yes")
    else:
        print("No")


solution()

# input
# 1 4
```

## B - Parking 2 (25:44)

`-` 무려 $5$번 만에 맞혔다. 뭘 빼먹었길래...

`-` $[L,R]$과 $[A,B]$가 겹치지 않는 경우, 서로 포함 관계에 있는 경우, 일부 겹치는 경우에 대해 요금을 계산하면 된다

```python
def solution():
    X, Y, L, R, A, B = map(int, input().split())
    if B <= L or A >= R:
        print(Y * (B - A))
        return
    if A <= L and B >= R:
        print(Y * (L - A + B - R) + X * (R - L))
        return
    if A >= L and B <= R:
        print(X * (B - A))
        return
    if A <= L and B <= R:
        print(Y * (L - A) + X * (B - L))
        return
    print(X * (R - A) + Y * (B - R))


solution()

# input
# 700 300 9 17 7 21
```

## C - Reverse Permutation (93:35)

`-` 문제 보고 어떻게 풀지 생각이 안 나서 D번을 먼저 풀고 왔다. 생각보다 간단하다

`-` $S$의 $\operatorname{o}$ 원소들의 인덱스를 $j_1,j_2, \dots, j_m$이라 하자

`-` 그럼 $A_{j_m+1}$부터 $A_N$까지는 기존과 그대로이다. 연산 후의 배열을 $B$라 할 때 $B_{j_m+1}, \dots, B_N$가 될 것이다

`-` 그리고 $A_{j_m}$부터 $A_{j_{m-1} + 1}$까지는 홀수번만(=$1$) 뒤집어지므로 $B_1,\dots, B_{j_m - j_{m-1}}$가 될 것이다

`-` $A_{j_{m-2}+1}$부터 $A_{j_{m-1}}$까지는 짝수번만(=$2$) 뒤집어지므로 $B_{j_m - j_{m-1} + j_{m-2} + 1},\dots, B_{j_m}$이 될 것이다

`-` 다시 $A_{j_{m-2}}$부터 $A_{j_{m-3} + 1}$까지는 홀수번만(=$3$) 뒤집어지므로 $B_{j_m - j_{m-1}+1},\dots, B_{j_m - j_{m-1}+ j_{m-2} - j_{m-3}}$이 될 것이다

`-` 이를 쉽게 처리하기 위해 $\operatorname{left}, \operatorname{mid}, \operatorname{right}$ 배열을 고려하자

`-` $\operatorname{right}$ 배열은 $A_{j_m+1}, \dots, A_N$과 같다

`-` $\operatorname{vector}$ 변수를 고려할 것인데 이는 원소를 $\operatorname{left}$ 뒤에 삽입할지 $\operatorname{mid}$ 뒤에 삽입할지를 가리킨다

`-` $\operatorname{vector} = 0$이면 $\operatorname{left}$ 뒤에 삽입하고 $1$이면 $\operatorname{mid}$ 뒤에 삽입하자. 초기 $\operatorname{vector}$는 $1$이다 

`-` $S_{j_m}$부터 $S_1$까지 순회하며 $S_i = \operatorname{o}$이면 $\operatorname{vector} = 1 - \operatorname{vector}$으로 변경하자

`-` 그리고 $\operatorname{vector}$에 맞춰 $i$를 배열에 삽입하면 된다

`-` 최종적으로 $B$는 슈도코드상 $\operatorname{concat}(\operatorname{left}, \operatorname{reverse}(\operatorname{mid}), \operatorname{right})$가 된다

```python
def solution():
    N = int(input())
    S = input().rstrip()
    j = None
    for i, s_i in enumerate(S, start=1):
        if s_i != "o":
            continue
        j = i
    if j is None:
        print(*list(range(1, N + 1)))
        return
    left = []
    mid = []
    right = list(range(j + 1, N + 1))
    vector = 1
    for i in range(j, 0, -1):
        s_i = S[i - 1]
        if s_i == "o":
            vector = 1 - vector
        if vector == 0:
            left.append(i)
        else:
            mid.append(i)
    answer = left + mid[::-1] + right
    print(*answer)


solution()

# input
# 5
# ooxoo
```

## D - X to Y (79:14)

`-` $\left\lfloor\frac{x}{K}\right\rfloor = y$를 만족하는 $y$는 하나뿐이다. 연산 $A$라 칭하겠다

`-` 반면 $\left\lfloor\frac{y}{K}\right\rfloor = x$를 만족하는 $y$의 범위는 $[Kx,\, Kx + K - 1]$이다. 연산 $B$라 칭하겠다

`-` 잘 생각해보면 연산 $B$를 수행한 뒤 연산 $A$를 수행하는 건 아무 의미가 없다. 연산 $B$를 수행하기 전의 $x$와 같아지기 때문이다

`-` 따라서 $X \to Y$ 과정을 생각해보면 연산 $A$만 수행하거나 (과정 $a$), 연산 $B$만 수행하거나 (과정 $b$), 연산 $A$만 수행하다가 바꿔서 연산 $B$만 수행하는 것이다 (과정 $c$)

`-` 이 $3$가지 중 정답이 있다

`-` 과정 $a$는 연산 $A$만 반복하면 되므로 결괏값을 얻기 쉽다. 반면 연산 $B$만 반복해서 $X$를 $Y$로 만들 수 있는지 어떻게 확인할 수 있을까?

`-` $x$를 $Kx$로만 만들기와 $Kx + K - 1$로만 만들기를 각각 고려하자. 만약 이를 진행하다가 전자는 $Y$ 이하이고 후자는 $Y$ 이상인 순간이 온다면 연산 $B$를 적절히 조정해서 $X \to Y$가 가능하다

`-` 이러한 순간이 오기 전에 전자가 $Y$를 초과해버린다면 $X \to Y$는 불가능하다

`-` 이제 과정 $c$만 확인하면 된다

`-` $1$번 연산은 $x$가 $0$이 된 이후에는 진행할 필요가 없다 (어차피 $0$임). 따라서 $1$번 연산은 최대 $O\left(\log_K{X}\right)$번만 가능하니 브루트 포스를 통해 결괏값을 찾자

`-` 과정 $a,b,c$를 통해 얻은 결괏값 중 최솟값이 정답이 된다

```python
def step_a(x, y, k):
    if x < y:
        return INF
    count = 0
    while x > y:
        x //= k
        count += 1
    if x == y:
        return count
    return INF


def step_b(x, y, k):
    if x > y:
        return INF
    count = 0
    p = q = x
    while p <= y:
        p = p * k
        q = q * k + k - 1
        count += 1
        if p <= y and q >= y:
            return count
    return INF


def step_c(x, y, k):
    min_count = INF
    count = 0
    while x > 0:
        x //= k
        count += 1
        b_count = step_b(x, y, k)
        min_count = min(count + b_count, min_count)
    return min_count


def solve_testcase(X, Y, K):
    c1 = step_a(X, Y, K)
    c2 = step_b(X, Y, K)
    c3 = step_c(X, Y, K)
    return min(c1, c2, c3)


def solution():
    global INF
    INF = float("inf")
    T = int(input())
    for _ in range(T):
        X, Y, K = map(int, input().split())
        answer = solve_testcase(X, Y, K)
        print(answer)


solution()

# input
# 1
# 11 9 3
```

`-` 에디토리얼을 봤는데 연산의 특성을 활용하여 매우 쉽게 해결한다

`-` 연산 $B$는 사실 $Y$에 연산 $A$를 사용하는 것과 같다

`-` 이제 연산 $A$만 존재하고 이를 $X$와 $Y$에 적용할 수 있다고 생각하자

`-` $X=Y$면 여태까지 적용한 연산 횟수가 정답이다. $X > Y$라면 $X$를 줄여야만 하므로 $X$에 연산 $A$를 적용하고 카운팅하자. $X < Y$라면 $Y$를 줄여야만 하므로 $Y$에 연산 $A$를 적용하고 카운팅하자

`-` 이를 반복하면 정답을 얻을 수 있다

## E - Digit Circus (Upsolving)

`-` $x$는 $3$의 배수여야 한다. 이를 추적하기 위해 여태까지 확정된 숫자들의 합을 $3$으로 나눈 나머지를 가지고 있자

`-` 여태까지 사용한 숫자 집합을 비트마스크로 가지고 있자. 이를 토대로 사용한 고유 숫자의 개수와 $3$의 사용 여부를 알 수 있다. 또한 선행 $0$ 여부도 알 수 있다

`-` 마지막으로 $N$ 이하 여부를 가지고 있자 ($N$보다 작지 않다면 $N$을 초과하면 안 되니 숫자 선택에 제약을 받는다)

`-` 앞에서부터 $i$번째 숫자까지 확정됐다고 하자. 그럼 알고 있는 정보는 다음과 같다

`-` 각 자리 숫자들의 합을 $3$으로 나눈 나머지, 사용한 숫자 집합, $N$ 이하 여부

`-` 만약 현재 선행 $0$이라면 여태까지 $0$을 사용했다는 의미이고 이는 그만큼 자릿수가 작다는 뜻이다. 따라서 $N$보다 작으며 사용한 숫자 집합에 $0$이 들어가지 않는다. 단, 일의 자리까지 선행 $0$일 수는 없다

`-` 각 상태 공간별로 수의 개수를 저장하고 있자. 그리고 토글링을 사용해 메모리 사용량을 줄이자

`-` 순회를 끝낸 뒤 $3$ 조건 중 하나만 만족하는 상태 공간에 저장된 수의 개수를 합산하면 정답이다

```python
from itertools import product


def digit_dp(n, mod):
    max_bit = 2**10 - 1
    n_digits = len(n)
    dp_prev = [[[0] * 2 for _ in range(3)] for _ in range(max_bit + 1)]
    dp_prev[0][0][0] = 1
    for i in range(n_digits):
        n_i = int(n[i])
        dp_curr = [[[0] * 2 for _ in range(3)] for _ in range(max_bit + 1)]
        for bitmask, remainder, is_less in product(range(max_bit + 1), range(3), range(2)):
            count = dp_prev[bitmask][remainder][is_less]
            if count == 0:
                continue
            stop = 9 if is_less else n_i
            is_leading_zero = bitmask == 0
            for d in range(stop + 1):
                next_bitmask = bitmask if is_leading_zero and d == 0 and i < n_digits - 1 else bitmask | (1 << d)
                next_remainder = (remainder + d) % 3
                next_is_less = is_less or d < n_i
                dp_curr[next_bitmask][next_remainder][next_is_less] += count
                dp_curr[next_bitmask][next_remainder][next_is_less] %= mod
        dp_prev = dp_curr
    return dp_prev


def solution():
    mod = 998244353
    N = input().rstrip()
    n_digits = len(N)
    max_bit = 2**10 - 1
    dp = digit_dp(N, mod)
    answer = 0
    for bitmask in range(2, max_bit + 1):
        has_3 = bool(bitmask & (1 << 3))
        n_unique = bin(bitmask).count("1")
        partial = (n_unique == 3) + has_3
        if partial > 1:
            continue
        for remainder in range(3):
            condition = partial + (remainder == 0)
            if condition != 1:
                continue
            answer += sum(dp[bitmask][remainder])
            answer %= mod
    print(answer)


solution()

# input
# 45
```

`-` Digit DP를 사용하여 해결한 첫 번째 문제인데 재밌는 알고리즘 같다
