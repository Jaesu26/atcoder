## A - Nine or Nein (1:22)

`-` 지문을 그대로 구현하자. 단, 실수 오차를 피하기 위해 $A \div B$가 $9$인지 확인하지 말고 $A = 9B$인지 확인하자

```python
def solution():
    A, B = map(int, input().split())
    if A + B == 9 or A - B == 9 or A * B == 9 or A == 9 * B:
        print("Nine")
        return
    print("Nein")


solution()

# input
# 16 9
```

## B - Survey Tabulation (3:56)

`-` 모든 문자를 소문자로 취급하자. 그리고 문자열별 빈도수를 계산한 뒤 최댓값을 출력하면 된다

```python
from collections import Counter


def solution():
    N = int(input())
    strings = [input().rstrip().lower() for _ in range(N)]
    answer = max(Counter(strings).values())
    print(answer)


solution()

# input
# 3
# abc
# AbC
# qwer
```

## C - Cookies and Greedy Takahashi (9:10)

`-` $0$을 기준으로 이보다 왼쪽에 있는 쿠키와 오른쪽에 있는 쿠키로 나눈 뒤 각각을 오름차순으로 정렬하자

`-` 왼쪽 쿠키는 스택으로 관리하고 오른쪽 쿠키는 큐로 관리하자. 그럼 현재 위치에서 가장 가까운 쿠키를 $O(1)$에 알 수 있고 쿠키 제거 및 위치 갱신도 $O(1)$에 할 수 있다

```python
from collections import deque


def solution():
    N = int(input())
    array = list(map(int, input().split()))
    array.sort()
    left = [a for a in array if a < 0]
    right = deque([a for a in array if a > 0])
    total = 0
    cx = 0
    while left and right:
        if cx - left[-1] <= right[0] - cx:
            nx = left.pop()
        else:
            nx = right.popleft()
        total += abs(cx - nx)
        cx = nx
    while left:
        nx = left.pop()
        total += cx - nx
        cx = nx
    while right:
        nx = right.popleft()
        total += nx - cx
        cx = nx
    print(total)


solution()

# input
# 4
# -1 -4 2 -11
```

## D - Chargers (20:52)

`-` 나이브한 방법은 $O\left(Q^2\right)$이므로 시간 초과이다. 배터리가 시간당 $1$씩 충전되는 걸 반영하기 위해 $1$번 쿼리가 들어오면 시점 $0$을 기준으로 해당 배터리의 충전량이 얼마였는지를 계산하여 배열에 추가하자. 이러면 기준이 통일되기에 배열의 원소들이 올바른 대소관계를 가진다

`-` 이제 $2$번 쿼리가 들어오면 배열에서 최댓값을 뽑은 뒤 $t_q$를 더해주면 된다. 이때 최댓값은 $V$를 넘길 수 없음에 유의하자

`-` 배열 대신 우선순위 큐를 사용하면 $O(\log Q)$에 최댓값을 뽑을 수 있다. 그럼 전체 알고리즘의 시간 복잡도는 $O(Q\log Q)$가 된다

```python
import heapq


def solution():
    Q, V = map(int, input().split())
    pq = []
    for _ in range(Q):
        query = list(map(int, input().split()))
        if query[0] == 1:
            _, t_q, w_q = query
            heapq.heappush(pq, t_q - w_q)
        else:
            if not pq:
                print(-1)
                continue
            _, t_q = query
            x = -1 * heapq.heappop(pq) + t_q
            print(min(x, V))


solution()

# input
# 2 3
# 1 5 0
# 2 7
```

## E - Sum of Square of Sum (Upsolving)

`-` 에디토리얼 봤다. 고등학생 때였으면 오히려 쉽게 풀었을 듯

`-` 선택된 $K$개의 수를 $x_1 \dots, x_K$라고 하자. $\textbf{score} = \left(\sum\limits_{i=1}^{K} x_i\right)^2 = (x_1 + \cdots + x_K)^2 = \sum\limits_{i=1}^{K} {x^2_i} + \sum\limits_{i\ne j}{x_i x_j}$

`-` $\sum \textbf{score}$에서 임의의 $A_i$에 대해 $A_i^2$이 몇 번 포함될까? $N$개 중 $K$개를 고를 건데 자기는 포함시켜야 하니 $N-1$개 중 $K-1$개를 고르면 된다. 따라서 $\binom{N-1}{K-1}$번 포함된다

`-` 비슷한 논리로 $A_i A_j$가 몇 번 포함되는지 구하자. $A_i$와 $A_j$를 고른 뒤 남은 $N-2$개에서 $K-2$개를 고르면 되니 경우의 수는 $\binom{N-2}{K-2}$이다. 이때 $\sum\limits_{i \ne j} A_i A_j = \sum A_i A_j - \sum\limits_{i} A_i^2$이다

`-` $\sum A_i A_j = (x_1x_1 + x_1x_2 + \cdots x_1x_K) + \cdots  + (x_Kx_1 + x_2x_2 + \cdots x_2x_K) = x_1 \sum\limits_{i=1}^{K} x_i + \cdots +  x_K \sum\limits_{i=1}^{K} x_i = \left(\sum\limits_{i=1}^{K} x_i\right)^2 $

`-` 따라서  $\sum\limits_{i \ne j} A_i A_j = \left(\sum\limits_{i} A_i\right)^2 - \sum\limits_{i} A_i^2$이다

`-` $A$의 모든 원소는 대칭적이다. 그러므로 $\sum \textbf{score} = \binom{N-1}{K-1} \cdot \sum\limits_{i=1}^{N} A_i^2 + \binom{N-2}{K-2} \cdot \left(\left(\sum\limits_{i=1}^{N} A_i\right)^2 - \sum\limits_{i=1}^{N} A_i^2\right)$

`-` 조합의 값을 계산하려면 모듈러 역원을 사용해야 한다. 따라서 전체 알고리즘의 시간 복잡도는 $O(N + \log \text{MOD})$이다

```python
def nCr(n, r, mod, factorials):
    if n < r or r < 0:
        return 0
    if r == 0:
        return 1
    if r == 1:
        return n % mod
    f = factorials
    return (f[n] * pow(f[r] * f[n - r], mod - 2, mod)) % mod


def solution():
    mod = 998244353
    N, K = map(int, input().split())
    A = list(map(int, input().split()))
    s1 = sum(A) % mod
    s2 = sum(map(lambda x: x**2, A)) % mod
    factorials = [1] * N
    for i in range(1, N):
        factorials[i] = (i * factorials[i - 1]) % mod
    bc1 = nCr(N - 1, K - 1, mod, factorials)
    bc2 = nCr(N - 2, K - 2, mod, factorials)
    answer = (bc1 * s2 + bc2 * (s1**2 - s2)) % mod
    print(answer)


solution()

# input
# 3 2
# 1 10 100
```

`-` $\left(\sum\limits_{i}x_i\right)^2 = \sum\limits_{i} {x^2_i} + \sum\limits_{i\ne j}{x_i x_j}$ 꼴로 표현하는 게 중요한 문제였다 (이걸 생각 못했음)
