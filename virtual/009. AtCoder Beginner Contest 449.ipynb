## B - Deconstruct Chocolate (6:55)

`-` $1$번 쿼리가 주어졌다고 해보자. 기본적으로 $RW$만큼 존재하나 여태까지 입력된 $2$번 쿼리에 의해 줄어든 너비를 차감하여 고려하면 된다

`-` $2$번 쿼리도 마찬가지로 처리하면 되며 전체 알고리즘의 시간 복잡도는 $O(Q)$이다. $H,W \le 100$이라 나이브하게 $O(HWQ)$에 풀 수도 있다

```python
def solution():
    H, W, Q = map(int, input().split())
    k1 = k2 = 0
    for _ in range(Q):
        op, t = input().split()
        t = int(t)
        if op == "1":
            print(t * (W - k2))
            k1 += t
        else:
            print(t * (H - k1))
            k2 += t


solution()

# input
# 7 9 5
# 2 4
# 1 3
# 2 1
# 2 1
# 1 3
```

## C - Comfortable Distance (35:47)

`-` $j$를 순회하며 가능한 $i$의 개수를 세자. $i$는 $j$로부터 최소 $L$만큼 떨어지고 최대 $R$만큼 떨어진 연속된 구간으로 표현된다. 투 포인터로 구간이 $1$칸씩 오른쪽으로 이동할 때마다 변하는 알파벳의 빈도수를 갱신하면 된다

```python
from collections import defaultdict


def solution():
    N, L, R = map(int, input().split())
    S = input().rstrip()
    counts = defaultdict(int)
    answer = 0
    for j in range(L, N):
        l, r = j - R, j - L
        counts[S[r]] += 1
        if l > 0:
            counts[S[l - 1]] -= 1
        answer += counts[S[j]]
    print(answer)


solution()

# input
# 6 2 4
# aabcba
```

## D - Make Target 2 (82:13)

`-` 나이브는 $O(4 \times 10^{12})$로 TLE이다

`-` 색상은 두 좌표 중 절댓값이 더 큰 값에만 영향을 받음에 주목하자. $x$를 고정시키고 $D \le y \le U$에 대해 검은 점의 개수를 셀 것이다. 이때 축을 기준으로 점의 분포가 대칭이므로 절댓값을 씌워서 생각하면 편하다

`-` $f(x, y)$를 $(x,1), (x,2), \dots, (x,y)$ 중 검은 점의 개수로 정의하자. $1 \le p \le x$에 대해 $\max(x, p) = x$이므로 이 구간 내 검은 점의 개수는 $\min(x,y) \cdot I(\text{$x$ is even})$이다. $x < p \le y$인 구간 내 검은 점의 개수는 $\left\lfloor\frac{y-x}{2}\right\rfloor + I(\text{$(y - x)$ is odd, $x + 1$ is even})$이다. 둘을 적절히 취합해 반환하면 된다

`-` $D$와 $U$에 관계에 따라 적절히 $f$를 활용하면 고정된 $x$에 대해 검은 점의 개수를 계산할 수 있다

`-` $D=U$이면 $I(\text{$\max(x,y)$ is even})$이 검은 점 개수이다. 둘 중 하나가 $0$이라면 $f(x, y) + I(\text{$x$ is even})$이 검은 점 개수이다 ($f$는 $y=0$일 때는 제외하고 계산하기에 따로 셈해줘야 함)

`-` $D > 0$이거나 $U < 0$이라면 구간 합 계산하는 것처럼 셈하자. 이때 $x$축은 구간에 속하지 않으니 무시하자. 이제 남은 경우는 $U > 0, D < 0$으로 각각에 대해 검은 점의 개수를 센 뒤 합하면 된다. 그리고 $x$축을 고려하자

`-` 고정된 $x$마다 검은 점의 개수를 $O(1)$에 셀 수 있으므로 전체 알고리즘의 시간 복잡도는 $O(2 \times 10^6)$이다

```python
def f(x, y):
    x, y = abs(x), abs(y)
    x_is_even = x % 2 == 0
    if y <= x:
        return y * x_is_even
    return x * x_is_even + (y - x) // 2 + ((y - x) % 2 == 1 and not x_is_even)


def solution():
    L, R, D, U = map(int, input().split())
    answer = 0
    for x in range(L, R + 1):
        x_is_even = x % 2 == 0
        if D == U:
            answer += x_is_even
        elif D == 0:
            answer += f(x, U) + x_is_even
        elif U == 0:
            answer += f(x, D) + x_is_even
        elif D > 0:
            answer += f(x, U) - f(x, D - 1)
        elif U < 0:
            answer += f(x, D) - f(x, U + 1)
        else:
            answer += f(x, D) + f(x, U) + x_is_even
    print(answer)


solution()

# input
# -4 3 1 3
```
