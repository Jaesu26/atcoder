## A - Closed interval (0:22)

`-` 정수 $L,R$에 대해 닫힌구간 $[L, R]$ 사이의 정수 개수는 $R-L+1$이다

```python
def solution():
    L, R = map(int, input().split())
    answer = R - L + 1
    print(answer)


solution()

# input
# 1 10
```

## B - Mapping (2:46)

```python
def print_yes_or_no(boolean):
    print("Yes" if boolean else "No")   


def solution():
    N, M = map(int, input().split())
    F = list(map(int, input().split()))
    n_unique = len(set(F))
    print_yes_or_no(n_unique == N)
    print_yes_or_no(n_unique == M)


solution()

# input
# 3 4
# 1 2 4
```

## C - Straw Millionaire (9:59)

`-` 지문이 애매한데 테케를 보면 $1$번 노드에서 도달할 수 있는 노드의 개수를 세면 되는 것 같다

```python
def dfs(graph, source):
    stack = [source]
    visited = set([source])
    while stack:
        u = stack.pop()
        for v in graph[u]:
            if v in visited:
                continue
            visited.add(v)
            stack.append(v)
    return len(visited)


def solution():
    N, M = map(int, input().split())
    graph = [[] for _ in range(N + 1)]
    for _ in range(M):
        A, B = map(int, input().split())
        graph[A].append(B)
    source = 1
    answer = dfs(graph, source)
    print(answer)


solution()

# input
# 5 5
# 1 2
# 2 3
# 3 4
# 2 4
# 5 2
```

## D - (xx) (36:10)

`-` 연산이 대칭적이므로 $A$에 연산을 적용하나 $B$에 적용하나 동일하다. 둘 다 가능한 모든 `(xx)`를 `xx`로 치환한 뒤 일치 여부를 확인하자 (정규화라 칭하겠다). 둘이 같으면 $A$를 $B$로 만들 수 있는 것이다 (정확한 증명은 모르겠음)

`-` 임의의 문자열 $S$를 정규화하는 방법을 정리하자. $S$를 순회하며 문자 종류에 따라 처리하면 된다

`-` `(`라면 이전 문자가 `(`여야 한다. 그렇지 않다면 여태까지 기록해둔 문자를 그대로 저장하자 (이는 앞으로 설명한 모든 문자에 대해 성립한다. 단, `(`인 경우 자기 자신은 제외하고 저장한다)

`-` `x`인 경우 이전 문자가 `(` 또는 `x`여야 한다. 이전 문자가 `x`인 경우 자신을 포함해 연속된 `x`는 최대 $2$개여야 한다

`-` `)`인 경우 이전 두 문자가 `xx`여야 한다. `xx` 앞에 `(`가 존재한다면 `(`와 `)`를 제거할 수 있다

`-` $S$를 $1$번만 순회하면 정규화를 할 수 있으므로 전체 알고리즘의 시간 복잡도는 $O(|A| + |B|)$이다

```python
from collections import defaultdict


def normalize(string):
    stack = []
    d = defaultdict(int)
    prev = ""
    for s in string:
        if s == "(":
            if prev == ")" or prev == "x":
                _push(stack, d)
            d["n_l"] += 1
            prev = s
            continue
        if s == "x":
            if prev == ")" or (prev == "x" and d["n_x"] >= 2):
                d["n_x"] += 1
                _push(stack, d)
            else:
                d["n_x"] += 1
            prev = s
            continue
        if prev == "x" and d["n_x"] == 2 and d["n_l"] > 0:
            d["n_l"] -= 1
            continue
        d["n_r"] += 1
        _push(stack, d)
        prev = s
    _push(stack, d)
    return "".join(stack)


def _push(stack, d):
    stack.extend(["("] * d["n_l"] + ["x"] * d["n_x"] + [")"] * d["n_r"])
    d.clear()


def solve_testcase(A, B):
    a = normalize(A)
    b = normalize(B)
    if a == b:
        return "Yes"
    return "No"


def solution():
    T = int(input())
    for _ in range(T):
        A = input().rstrip()
        B = input().rstrip()
        answer = solve_testcase(A, B)
        print(answer)


solution()

# input
# 1
# ((xx)xx)xx
# (x((xx))x)(xx)
```

`-` 이렇게 구현하는 게 맞나 싶어 에디토리얼 봤는데 당연히 아니었다. 내가 정규화라 칭한 건 문자열을 [canonical form](https://en.wikipedia.org/wiki/Canonical_form)으로 만드는 것과 같다. 따라서 스택을 이용해 최근 추가된 문자 $4$개가 `(xx)`이면 `xx`로 치환하면 된다 (나처럼 어렵게 할 필요가 없음)

`-` 이와 유사한 문제는 연산의 특성을 확인한 뒤 canonical form을 생각하며 풀면 될 것 같다
