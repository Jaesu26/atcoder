## A - Grapes (1:53)

`-` $M = qN + r$일 때 기본적으로 모두에게 $q$개의 포도를 줄 수 있다. 그리고 앞의 $r$명에겐 추가로 $1$개의 포도를 줄 수 있다

```python
def solution():
    N, M = map(int, input().split())
    q = M // N
    r = M % N
    answer = [q + 1] * r + [q] * (N - r)
    print("\n".join(map(str, answer)))


solution()

# input
# 3 5
```

## B - Topping (10:05)

`-` 냅색 DP에 추가로 여태까지 선택한 토핑 개수를 고려하자. 그럼 $O(3NV)$에 해결할 수 있다

`-` 콘테스트 중엔 더 쉬운 풀이를 찾다 시간을 내다 버렸다 (이게 B라고?)

`-` 에디토리얼 봤는데 그냥 브루트 포스로 $O\left(N^3\right)$에 풀 수 있다;;; 

`-` 문제를 중간까지 읽었을 땐 배낭 문제구나 생각했는데 마지막에 $3$개만 고를 수 있다는 제약을 둬서 나이브 접근을 생각 못한 것 같다. 배낭 문제를 기저로 둔 상태에서 제약을 적용하다보니 떠올리지 못했음. 배낭 문제 변형인가? 이런 생각만 하다 시간 버림...

`-` 사실 토핑 가격이 단순 $i$인 것부터 쎄하긴 했다. 문제를 다 읽었으면 정리하는 타임을 한 번 가지자

```python
def solution():
    N, V = map(int, input().split())
    W = list(map(int, input().split()))
    dp = [[0] * (V + 1) for _ in range(4)]
    answer = []
    for i in range(1, N + 1):
        w = W[i - 1]
        for n in range(2, -1, -1):
            for t in range(i, V + 1):
                dp[n + 1][t] = max(dp[n][t - i] + w, dp[n + 1][t])
    answer = max(dp[3])
    print(answer)


solution()

# input
# 5 9
# 31 41 59 26 53
```

## C - Sort Subarray (37:29)

`-` 임의의 $(i,j)$에 대해 $A_i > A_j$라면 $A_i, A_j$를 포함하는 부분 배열을 정렬해야 한다. 이때 부분 배열의 길이가 $K$를 초과하면 $A$를 오름차순으로 만들 수 없다

`-` 임의의 $i$에 대해 $\max\limits_{1\le j < i}\{A_j\} > A_i$ 또는 $\min\limits_{i <  j \le N}\{A_j\} < A_i$를 만족한다면 정렬이 필요한 부분 배열에 $A_i$가 속해야 한다. 이는 세그먼트 트리를 사용하면 $O(\log N)$에 확인할 수 있으며 가능한 $i$는 $N$개이므로 총 $O(N \log N)$이다

`-` 최종적으로 정렬이 필요한 인덱스 중 최댓값과 최솟값의 차이가 $K$ 이상이라면 정답은 No이고 아니라면 Yes이다

```python
from atcoder.segtree import SegTree


def solution():
    N, K = map(int, input().split())
    A = list(map(int, input().split()))
    max_tree = SegTree(max, 0, A)
    min_tree = SegTree(min, N + 1, A)
    indices = []
    for i, a in enumerate(A):
        mx = max_tree.prod(0, i)
        mn = min_tree.prod(i, N)
        if a < mx or a > mn:
            indices.append(i)
    if indices and max(indices) - min(indices) >= K:
        print("No")
        return
    print("Yes")


solution()

# input
# 3 1
# 3 2 1
```

`-` 처음엔 세그먼트 트리 말고 누적 최솟값과 누적 최댓값을 이용해서 풀려고 했는데 잘 안되서 그냥 세그먼트 트리를 사용했다. 다음부턴 바로 세그먼트 트리를 쓰자

`-` 세그먼트 트리 템플릿이 있지만 코드를 깔끔하게 하려고 앳코더에서 제공하는 라이브러리를 사용했다

## D - Range Set Insertion Query (61:59)

`-` 나이브하게 구현하면 $O(NQ)$라 실제로 집합에 원소를 넣으면 안 된다. 집합마다 어떤 원소가 존재하는지는 필요 없고 원소의 개수만 중요하다

`-` 쿼리를 $X$ 기준으로 그룹 짓자. 원소를 중복해서 추가하면 안 되니 각 그룹마다 라인 스위핑을 통해 연산 적용 구간을 세팅하자. 그리고 구간마다 원소가 $1$개 추가됨을 마킹하면 된다

`-` 구간 업데이트를 빠르게 수행하기 위해 레이지 세그먼트 트리를 사용하면 문제를 $O(Q\log Q + (N + Q) \log N)$에 해결할 수 있다

```python
from itertools import groupby
from atcoder.lazysegtree import LazySegTree


def line_sweeping(lines):
    lines.sort()
    ranges = [lines.pop(0)]
    for l, r in lines:
        if l <= ranges[-1][1]:
            ranges[-1][1] = max(r, ranges[-1][1])
            continue
        ranges.append([l, r])
    return ranges


def op(u, v):
    return u[0] + v[0], u[1] + v[1]


def apply(lz, u):
    return u[0] + lz * u[1], u[1]


def compose(old, new):
    return old + new


def solution():
    N, Q = map(int, input().split())
    queries = [list(map(int, input().split())) + [i] for i in range(Q)]
    queries.sort(key=lambda q: q[2])
    ranges = []
    groups = groupby(queries, key=lambda q: q[2])
    for _, group in groups:
        lines = [[l, r] for l, r, _, _ in group]
        ranges.extend(line_sweeping(lines))
    tree = LazySegTree(op, (0, 1), apply, compose, 0, N + 1)
    for l, r in ranges:
        tree.apply(l, r + 1, 1)
    answers = [tree.prod(i, i + 1)[0] for i in range(1, N + 1)]
    print(*answers)


solution()

# input
# 8 5
# 2 5 1
# 1 4 2
# 7 8 1
# 3 6 3
# 2 5 2
```

`-` 레이지 세그먼트 트리 재귀로 구현해도 AC 받는다
